```python
"""
       In LangChain, we have document loaders. They load files like PDF, DOCX, CSV, TXT, HTML, and JSON.
       Internally, they may use parser libraries like pypdf, docx2txt, unstructured, or jq to extract the
       content. Then LangChain converts that extracted content into Document(page_content, metadata) 
       format for the RAG pipeline.

       In LangChain, parsed content comes in the form of a Document object.

       Inside this object:

                page_content = the actual extracted text
                metadata = source information like file name, page number, row number, etc.
"""

```

```python
"""

                 Python code to load pdf content using langchain pdf loader.

"""



from langchain_community.document_loaders import PyPDFLoader

pdf_path_file = r"D:\RAG\DataParsing\doc\document.pdf"
pdf_loader = PyPDFLoader(pdf_path_file)

"""
            1) PyPDFLoader usually does not load the entire PDF into pdf_docs[0].
               It usually loads the PDF page by page.

            2) Langchain will treat 1 Document object per PDF page.

            3) So if your PDF has 3 pages, then:

                    pdf_docs[0] → page 1 content
                    pdf_docs[1] → page 2 content
                    pdf_docs[2] → page 3 content                   
"""

pdf_docs = pdf_loader.load()          # Now pdf_docs is a list of Document objects, one for each page in the PDF.
pdf_metadata = pdf_docs[0].metadata
pdf_content = pdf_docs[0].page_content
#print(pdf_metadata)
print(pdf_content)
```

```python
"""

                 Python code to load text content using langchain text loader.

"""

from langchain_community.document_loaders import TextLoader

text_path = r"D:\RAG\DataParsing\doc\sample_text.txt"
text_loader = TextLoader(text_path)


"""
           TextLoader treats the whole text file as one single document.
"""

text_doc = text_loader.load()
text_metadata = text_doc[0].metadata
text_content = text_doc[0].page_content
print(text_metadata)
print(text_content)
```

```python
"""
           python code to load the CSV content using langchain CSV loader
"""

from langchain_community.document_loaders import CSVLoader

csv_path = r"D:\RAG\DataParsing\doc\sample_data.csv"
csv_loader = CSVLoader(csv_path)
csv_docs = csv_loader.load()
print("Total Number of rows loaded from csv: ", len(csv_docs))
print("-" * 50)

"""
      csv_doc becomes a list of Document objects — usually one Document per CSV row.
"""

for csv_doc in csv_docs:
    print("Metadata :", csv_doc.metadata)
    print("Content : ", csv_doc.page_content)
    print("-" * 50)

```

```python
"""
                    python code to load JSON content using Langchain JSON Loader
"""

from langchain_community.document_loaders import JSONLoader
json_path = r"D:\RAG\DataParsing\doc\sample_data.json"


"""
        1) jq_schema=".[]"

                if your JSON file contains an array of JSON objects, then jq_schema=".[]" returns 
                each JSON object one at a time and create documents from them, So it works similarly 
                to CSVLoader row by row.


        2) When you set text_content=False: 
        
               you are telling JSONLoader that my jq_schema may return structured JSON objects 
               (dictionaries or lists), not plain text. Please convert those objects into a text 
               representation before storing them in page_content.


        3) When you set text_content=True:


            you are telling JSONLoader that my jq_schema may return plain text but not structured JSON 
            objects (dictionaries or lists), Since the extracted value is already text, use it directly 
            as the page_content of the Document without performing any conversion.

"""


json_loader = JSONLoader(json_path, jq_schema=".[]",  text_content=False)
json_docs = json_loader.load()
for json_doc in json_docs:
    print("Metadata: ", json_doc.metadata)
    print("Content: ", json_doc.page_content)

```

```python
"""
                    python code to load HTML content using Langchain Unstructured HTML Loader
"""


from langchain_community.document_loaders import BSHTMLLoader
html_path = r"D:\RAG\DataParsing\doc\sample_page.html"


"""
      UnstructuredHTMLLoader loads the whole HTML content at once into a single Document by default.
      it won't extract title of html page.
"""


unstructured_html_loader = BSHTMLLoader(html_path)
html_content = unstructured_html_loader.load()
print("Metadata: ", html_content[0].metadata)
print("Content :", html_content[0].page_content)
```

```python
"""
         python code to load docx content using Langchain Docx2txtLoader
"""

from langchain_community.document_loaders import Docx2txtLoader

docx_path = r"D:\RAG\DataParsing\doc\sample_document.docx"

"""
         By default, Docx2txtLoader loads the entire .docx file into a single Document.
"""

docx_2_txt_loader_object = Docx2txtLoader(docx_path)
docx_content = docx_2_txt_loader_object.load()
print("Metadata: ", docx_content[0].metadata)
print("Content :", docx_content[0].page_content)
```

```python
"""
         python code to load .md file content using Langchain UnstructuredMarkdownLoader
"""

from langchain_community.document_loaders import UnstructuredMarkdownLoader

md_file_path = r"D:\RAG\DataParsing\doc\sample.md"


"""
         By default, UnstructuredMarkdownLoader loads the entire Markdown (.md) 
         file into a Document object.
"""

md_file_loader_object = UnstructuredMarkdownLoader(md_file_path)
md_file_loader = md_file_loader_object.load()

print(md_file_loader[0].metadata)
print(md_file_loader[0].page_content)
```

```python
"""
      python code to parse web contect for the given URL using Langchain WebBaseLoader
"""

from langchain_community.document_loaders import WebBaseLoader

url = "https://sunnysavita10.github.io/"


"""
     WebBaseLoader is a LangChain document loader that loads the text content from one or more web pages 
     (URLs) and converts each web page into a Document object. Internally, it downloads the HTML page, 
     parses it (using BeautifulSoup by default), extracts the visible text, and stores useful metadata 
     such as the page title and URL.

     WebBaseLoader fetches the HTML returned by the server. If a website renders most of its content using
     JavaScript (common in many modern single-page applications), the loader may return incomplete or 
     empty content because it doesn't execute JavaScript.
"""

web_loader_object = WebBaseLoader(url)
web_content = web_loader_object.load()
print("Metadata: ", web_content[0].metadata)
print("Content: ", web_content[0].page_content)
```
